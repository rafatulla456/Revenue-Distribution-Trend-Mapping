# Exploratory Data Analysis

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender',data = df)

# To indicate bar labels
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
df.groupby(['Gender'],as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending=False)
```

```python

sales_gen = df.groupby(['Gender'],as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending=False)
sns.barplot(x = 'Gender', y = 'Amount', data = sales_gen)
plt.show()
```

#### Age


```python
df.columns
```

```python
# hue is used to specify men and women in graph
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
# Total Amount vs Age Group 
sales_age = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group', y = 'Amount',data = sales_age)
plt.show()
```

From the above graph we can see that most of the buyers are of age group between 26-35 yrs female

## State

```python
# Total number of orders from to 10 states
sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by = 'Orders', ascending= False).head(10)


#Chart size
sns.set(rc={'figure.figsize':(15,5)})

sns.barplot(data = sales_state, x = 'State', y = 'Orders')
plt.show()
```

```python
# total amount/ sales from top 10 states
sales_state = df.groupby(['State'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize':(15,5)})

sns.barplot(data = sales_state, x = 'State', y = 'Amount')

plt.show()
```

#### Marital status

```python
ax = sns.countplot(data = df, x = 'Marital_Status')

sns.set(rc={'figure.figsize':(10,5)})
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
sales_stat = df.groupby(['Marital_Status','Gender'], as_index = False )['Amount'].sum().sort_values(by = 'Amount', ascending=False)

sns.set(rc = {'figure.figsize':(8,4)})
sns.barplot(data = sales_stat, x = 'Marital_Status', y = 'Amount', hue = 'Gender')
plt.show()
```

### Occupation

```python
sns.set(rc={'figure.figsize':(17,9)})

ax = sns.countplot(data = df, x = 'Occupation')

for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
# Occupation by Amount

sales_state = df.groupby(['Occupation'], as_index = False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.set(rc = {'figure.figsize':(17,6)})
sns.barplot(data = sales_state, x = 'Occupation', y = 'Amount')
plt.show()
```

#### Product category

```python
sns.set(rc = {'figure.figsize':(20,9)})
ax = sns.countplot(data = df, x = 'Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
# Product Vs Sales

prod_sales = df.groupby(['Product_Category'], as_index =  False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)
sns.set(rc = {'figure.figsize': (18,9)})

sns.barplot(data = prod_sales, x = 'Product_Category', y = 'Amount')
plt.show()
```

```python
#Top products

top_prod = df.groupby(['Product_ID'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.set(rc = {'figure.figsize':(18,6)})
sns.barplot(data = top_prod, x = 'Product_ID', y = 'Orders')
plt.show()
```
